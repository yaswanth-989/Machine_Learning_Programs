```python
import warnings
warnings.filterwarnings("ignore")

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.cluster import KMeans, DBSCAN, AgglomerativeClustering
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score
from sklearn.decomposition import PCA
from sklearn.neighbors import NearestNeighbors
from scipy.cluster.hierarchy import linkage, dendrogram
from scipy.spatial.distance import pdist

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", 100)
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.info()
```

```python
# Numerical and categorical columns
numeric_cols_all = df.select_dtypes(include=np.number).columns.tolist()
categorical_cols = df.select_dtypes(exclude=np.number).columns.tolist()

print("Numerical columns:")
print(numeric_cols_all)

print("\nCategorical columns:")
print(categorical_cols)

print("\nMissing values:")
display(df.isna().sum().sort_values(ascending=False).to_frame("missing_count").head(20))
```

```python
EXCLUDE_FROM_UNSUPERVISED = [
    "StudentID",
    "PlacementStatus",
    "Salary Package",
    "IsAnomaly"
]

missing_excluded = [c for c in EXCLUDE_FROM_UNSUPERVISED if c not in df.columns]
if missing_excluded:
    print("Columns not found (safe to ignore):", missing_excluded)

available_excluded = [c for c in EXCLUDE_FROM_UNSUPERVISED if c in df.columns]
print("Excluded from clustering:", available_excluded)

# Useful audit: relationship of candidate numeric variables with target.
if "PlacementStatus" in df.columns:
    corr = df[numeric_cols_all].corr(numeric_only=True)["PlacementStatus"].sort_values(ascending=False)
    display(corr.to_frame("corr_with_PlacementStatus"))
```

```python
FEATURES = [
    "SGPA_Sem1", "SGPA_Sem2", "SGPA_Sem3", "SGPA_Sem4",
    "SGPA_Sem5", "SGPA_Sem6", "SGPA_Sem7", "SGPA_Sem8",
    "CGPA", "AttendancePercent",
    "Internships", "Projects", "Workshops", "Certifications",
    "Publications", "AptitudeTestScore", "SoftSkillsRating",
    "CodingTestScore", "MockInterviewScore", "ExtraCurricular"
]

FEATURES = [c for c in FEATURES if c in df.columns]

print("Number of modeling features:", len(FEATURES))
print(FEATURES)

X_raw = df[FEATURES].copy()
y = df["PlacementStatus"].copy() if "PlacementStatus" in df.columns else None

display(X_raw.describe().T)
```

```python
# Exploratory preprocessing for the unsupervised experiments.
# For the final "cluster lift" experiment later, we fit these transformations on training data only.

preprocess = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

X_scaled = preprocess.fit_transform(X_raw)

X_scaled = pd.DataFrame(X_scaled, columns=FEATURES, index=df.index)

print("Processed matrix shape:", X_scaled.shape)
display(X_scaled.head())

print("\nMeans after standardisation (approximately 0):")
display(X_scaled.mean().round(3).to_frame("mean"))

print("\nStd after standardisation (approximately 1):")
display(X_scaled.std(ddof=0).round(3).to_frame("std"))
```

```python
# Optional visual check: PCA projection of the standardized feature space.
# This is only for visualization; K-Means itself uses the full feature space.

pca_vis = PCA(n_components=2, random_state=RANDOM_STATE)
X_pca2 = pca_vis.fit_transform(X_scaled)

plt.figure(figsize=(8, 6))
plt.scatter(X_pca2[:, 0], X_pca2[:, 1], s=8, alpha=0.35)
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("Student Data Projected onto First Two Principal Components")
plt.show()

print("Variance explained by PC1 + PC2:",
      round(pca_vis.explained_variance_ratio_.sum() * 100, 2), "%")
```

```python
K_VALUES = list(range(2, 9))

results_kmeans = []

for k in K_VALUES:
    km_random = KMeans(
        n_clusters=k,
        init="random",
        n_init=10,
        random_state=RANDOM_STATE
    )
    labels_random = km_random.fit_predict(X_scaled)

    km_plus = KMeans(
        n_clusters=k,
        init="k-means++",
        n_init=10,
        random_state=RANDOM_STATE
    )
    labels_plus = km_plus.fit_predict(X_scaled)

    results_kmeans.append({
        "K": k,
        "Random_Inertia": km_random.inertia_,
        "Random_Silhouette": silhouette_score(X_scaled, labels_random),
        "KMeans++_Inertia": km_plus.inertia_,
        "KMeans++_Silhouette": silhouette_score(X_scaled, labels_plus)
    })

kmeans_results = pd.DataFrame(results_kmeans)
display(kmeans_results.round(4))
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(kmeans_results["K"], kmeans_results["Random_Inertia"], marker="o", label="Random init")
ax.plot(kmeans_results["K"], kmeans_results["KMeans++_Inertia"], marker="o", label="K-Means++")
ax.set_xlabel("Number of clusters (K)")
ax.set_ylabel("Inertia")
ax.set_title("Elbow Method: Inertia vs K")
ax.legend()
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(9, 5))
ax.plot(kmeans_results["K"], kmeans_results["Random_Silhouette"], marker="o", label="Random init")
ax.plot(kmeans_results["K"], kmeans_results["KMeans++_Silhouette"], marker="o", label="K-Means++")
ax.set_xlabel("Number of clusters (K)")
ax.set_ylabel("Silhouette score")
ax.set_title("Silhouette Score vs K")
ax.axhline(0, linewidth=0.8)
ax.legend()
plt.show()

best_k_silhouette = int(kmeans_results.loc[
    kmeans_results["KMeans++_Silhouette"].idxmax(), "K"
])
print("K with highest K-Means++ silhouette:", best_k_silhouette)
```

```python
K_BEST = best_k_silhouette

km_random_final = KMeans(
    n_clusters=K_BEST,
    init="random",
    n_init=10,
    random_state=RANDOM_STATE
)
labels_random_final = km_random_final.fit_predict(X_scaled)

km_plus_final = KMeans(
    n_clusters=K_BEST,
    init="k-means++",
    n_init=10,
    random_state=RANDOM_STATE
)
labels_plus_final = km_plus_final.fit_predict(X_scaled)

comparison = pd.DataFrame({
    "Model": ["K-Means random", "K-Means++"],
    "K": [K_BEST, K_BEST],
    "Inertia": [km_random_final.inertia_, km_plus_final.inertia_],
    "Silhouette": [
        silhouette_score(X_scaled, labels_random_final),
        silhouette_score(X_scaled, labels_plus_final)
    ],
    "Davies_Bouldin": [
        davies_bouldin_score(X_scaled, labels_random_final),
        davies_bouldin_score(X_scaled, labels_plus_final)
    ]
})

display(comparison.round(4))
```

```python
# PCA visualization of K-Means++ clusters
plt.figure(figsize=(9, 6))
plt.scatter(
    X_pca2[:, 0], X_pca2[:, 1],
    c=labels_plus_final, s=10, alpha=0.45, cmap="tab10"
)
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title(f"K-Means++ Clusters (K={K_BEST}) shown in PCA 2-D space")
plt.colorbar(label="Cluster")
plt.show()
```

```python
profile_df = df[FEATURES].copy()
profile_df["Cluster"] = labels_plus_final

cluster_profile = profile_df.groupby("Cluster")[FEATURES].mean().round(2)
cluster_sizes = pd.Series(labels_plus_final).value_counts().sort_index().rename("Student_Count")

display(pd.concat([cluster_sizes, cluster_profile], axis=1))
```

```python
# Heatmap of standardized cluster profiles.
cluster_profile_scaled = cluster_profile.copy()

# Standardize each feature across clusters only for visualization.
heatmap_data = (
    cluster_profile_scaled - cluster_profile_scaled.mean()
) / cluster_profile_scaled.std(ddof=0).replace(0, 1)

plt.figure(figsize=(14, 7))
sns.heatmap(heatmap_data.T, center=0, cmap="coolwarm")
plt.xlabel("Cluster")
plt.ylabel("Feature")
plt.title("Cluster Profiles — Relative Strength by Feature")
plt.show()
```

DBSCAN

```python
# DBSCAN can be computationally expensive on 50k observations.
# For parameter exploration we use a reproducible sample.
DBSCAN_SAMPLE_SIZE = min(10000, len(X_scaled))
dbscan_rng = np.random.default_rng(RANDOM_STATE)
dbscan_idx = dbscan_rng.choice(len(X_scaled), size=DBSCAN_SAMPLE_SIZE, replace=False)
X_db_sample = X_scaled.iloc[dbscan_idx].to_numpy()

print("DBSCAN parameter-search sample:", X_db_sample.shape)

EPS_VALUES = [0.8, 1.0, 1.2, 1.4, 1.6, 1.8, 2.0]
MIN_SAMPLES_VALUES = [5, 10, 20]

dbscan_results = []

for eps in EPS_VALUES:
    for min_samples in MIN_SAMPLES_VALUES:
        db = DBSCAN(eps=eps, min_samples=min_samples, n_jobs=-1)
        labels = db.fit_predict(X_db_sample)

        n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
        noise_pct = (labels == -1).mean() * 100

        non_noise = labels != -1
        if n_clusters >= 2 and non_noise.sum() > 1:
            sil = silhouette_score(X_db_sample[non_noise], labels[non_noise])
            dbi = davies_bouldin_score(X_db_sample[non_noise], labels[non_noise])
        else:
            sil = np.nan
            dbi = np.nan

        dbscan_results.append({
            "eps": eps,
            "min_samples": min_samples,
            "clusters": n_clusters,
            "noise_pct": noise_pct,
            "silhouette_non_noise": sil,
            "davies_bouldin_non_noise": dbi
        })

dbscan_results = pd.DataFrame(dbscan_results)
display(dbscan_results.sort_values(
    ["silhouette_non_noise", "noise_pct"],
    ascending=[False, True]
).round(4))
```

```python
# Visualize DBSCAN sensitivity to eps for each min_samples value.
fig, ax = plt.subplots(figsize=(10, 6))

for m in MIN_SAMPLES_VALUES:
    sub = dbscan_results[dbscan_results["min_samples"] == m]
    ax.plot(sub["eps"], sub["noise_pct"], marker="o", label=f"min_samples={m}")

ax.set_xlabel("eps")
ax.set_ylabel("Noise percentage")
ax.set_title("DBSCAN Sensitivity — Noise Percentage")
ax.legend()
plt.show()
```
